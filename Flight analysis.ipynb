```python
#highest departures
#earlymorning, morning, afternoon, evening, night, Latenight

def flight_dep(x):
    
    if x>4 and x<=8:
        return 'EarlyMorning'
        
    elif x>8 and x<=12:
        return 'Morning'
        
    elif x>12 and x<=16:
        return 'Afternoon'
        
    elif x>16 and x<=20:
        return 'Evening'
        
    elif x>20 and x<=24:
        return 'Night'
        
    else:
        return 'LateNight'
```

```python
#apply
flights['departure_hours'].apply(flight_dep).value_counts().plot(kind = 'bar')
```

```python
#IQR METHOD
Q1 = np.quantile(flights['Price'], 0.25)
Q2 = np.quantile(flights['Price'], 0.50)
Q3 = np.quantile(flights['Price'], 0.75)

IQR = Q3 - Q1

HF = Q3 + 1.5 * IQR
LF = Q1 - 1.5 * IQR



outliers = [i for i in flights['Price'] if i < LF or i > HF]

```
